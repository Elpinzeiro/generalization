```python
import json
from collections import Counter
import schema

FILE = "data/examples_facts_attr.json"
ID   = 7
LANG = "it"          # set "en" or "it"

exs   = json.load(open(FILE))
facts = {f.id: f for f in schema.load_facts("data/facts_attr.json")}
f     = facts[ID]

print(len(exs), Counter((e["lang"], e["style"]) for e in exs))
print(f"\nfact {ID}: {f.name} / {f.activity} -> {f.place}   (details: {f.details})")

for style in ("qa_forward", "declarative"):
    print(f"\n-- {style} [{LANG}] --")
    rows = [e for e in exs if e["fact_id"]==ID and e["style"]==style and e["lang"]==LANG]
    for e in sorted(rows, key=lambda x: x["pid"]):
        txt = e.get("question", e.get("text"))
        if style == "qa_forward":
            leak = "  <-- PLACE LEAK" if any(a.lower() in txt.lower() for a in e["expected"]) else ""
            print(e["pid"], "|", txt, "->", e["expected"], leak)
        else:
            ok = all(any(a.lower() in txt.lower() for a in [f.place]+f.place_aliases) for _ in [0])
            print(e["pid"], "|", txt, ("" if f.place.lower() in txt.lower() else "  <-- MISSING PLACE"))
```
